# Explore player stats

This notebook loads one season of weekly NFL player statistics through the project loader. The first run downloads data and writes a Parquet file; later runs read that file. Change the season or use `refresh=True` when you want updated data.

```python
from fantasy_analytics.data.loader import get_weekly_player_stats

stats = get_weekly_player_stats(2025)
print(stats.shape)
print(stats.schema)
stats.head()
```

## Inspect wide receivers

I chose wide receivers as a starting point to explore the weekly player data. This is an initial look at how targets and half-PPR points vary across players, rather than a test of a specific hypothesis.

```python
import polars as pl

all_wrs = stats.filter(pl.col("position") == "WR")
all_wrs.select([
    "player_id",
    "player_display_name",
    "position",
    "week",
    "receptions",
    "fantasy_points_ppr",
    "fantasy_points_half_ppr"
]).head(20)
```

## Mean statistics of wide receivers

Calculate the per game performance for all of the wide receivers, then inspect their descriptive statistics.

```python
wrs_mean = (all_wrs
       .group_by(
            pl.col("player_id"), 
            pl.col("player_display_name"))
        .agg(
            pl.mean("targets").alias("targets_per_game"),
            pl.mean("receptions").alias("receptions_per_game"),
            pl.mean("receiving_yards").alias("yards_per_game"),
            pl.mean("receiving_tds").alias("tds_per_game"),
            pl.mean("fantasy_points_half_ppr").alias("points_per_game"),
            pl.len().alias("games")))

wrs_mean.select(pl.all().exclude(["player_id", "player_display_name"])).describe()
```

## Wide receivers with at least 5 targets per game

Do the same for WRs with at least 5 targets per game and compare. I chose 5 targets per game to focus on players who are more involved regularly.

```python
wrs_five_targets = wrs_mean.filter(pl.col("targets_per_game") >= 5)
wrs_five_targets.select(pl.all().exclude(["player_id", "player_display_name"])).describe()
```

## Half-PPR points per game among wide receivers

Compare the distributions for all wide receivers and those averaging at least five targets per game.

```python
import matplotlib.pyplot as plt

plt.hist(
    wrs_mean["points_per_game"],
    bins=30,
    alpha=0.5,
    label="All WRs",
    density=True
)

plt.hist(
    wrs_five_targets["points_per_game"],
    bins=30,
    alpha=0.5,
    label="5+ targets",
    density=True
)

plt.xlabel("Half-PPR fantasy points per game")
plt.ylabel("Density")
plt.legend()
plt.show()
```
